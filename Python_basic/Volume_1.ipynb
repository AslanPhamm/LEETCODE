## **Change variable type**

### **int(variable)** to convert to integer

**int(variable)** function only convert string that contains only digits, no decimal point, no letters, no symbols.

```python
x=int("23")
print(x)
y=int(23.45)
print(y)
z=int(23+45)
print(z)
```

**int(variable)** can't convert string float to integer or string fomula to integer

```python
int("23.45")  # Error
int("23abc")  # Error
int("34+5")   # Error
```

```python
# To convert int variable "23.45", we need to convert it to float first, then convert to int
a= float(23.45)
b= int(a)
print(b)
```

### **float(variable)** to convert to float
**float(variable)** function only convert string that contains digits and decimal point, no letters, no symbols.

```python
x= float("23.45")
print(x)
y= float(23)
print(y)
z= float(23+45+67.89)
print(z)
```

**float(variable)** can't convert string fomula to float

```python
float("23+45")  # Error
float("23abc")  # Error
```

### **str(variable)** to convert to string
**str(variable)** function convert any variable to string, including integer, float, boolean, list, tuple, dictionary, set, etc.

```python
x=str(23)
print(x)
y=str(23+45+67.89)
print(y)
```

**str(variable)** can't convert data string (1,2,3,...)

```python
str(2,3)  # Error
str([1,2,3])  # Error
```

### **eval(variable)** convert string fomula to integer or float
**eval(variable)** can recognize and evaluate string fomula, including +, -, *, /, **, %, //, (), etc in the same line, and break whith comma (,)

```python
x=eval("23+45")
print(x)
y=eval("15**2+7.0+13*(34-2**1.5)")
print(y)
z=eval("1+2, 3**2, 14.2+16.5**2")
print(z)
m,n=eval("1+2, 3**2")
m,n
```

## **if statement**

**if** <condition>:
    <code runs if condition is True>
**elif** <another_condition>:
    <code runs if this condition is True>
**else:**
    <code runs if none of the above is True>


- if → checks the first condition.

- elif (else if) → checks another condition if the first was False.

- else → runs if all conditions are False.

Indentation (spaces at the beginning of the line) is required in Python.
