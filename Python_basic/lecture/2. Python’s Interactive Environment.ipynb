# **1. Python Data types**

## **a. The `type()` Function and Main Data Types in Python**

The ```type(<data>)``` function returns the type of the provided ```<data>```.


````python
type(2.4)
````

````python
type("Hochiminh")
````

````python
type(3>2)
````

The ```type()``` function can return the following common data types in Python:
- ```int```: Integer numbers.

- ```bool```: Boolean values, which can be either ```True``` or ```False```.
- ```str```: String, a sequence of characters.

## **b. Variables Do Not Require Declaration or Type Definition**
In Python, variables do not require explicit declaration or type definition. The type is automatically inferred based on the assigned value.
See the program below:

````python
x
````

````python
x=1
````

````python
x
````

In the first line, when ```x``` is used, Python will not recognize it because ```x``` has not been assigned a value or defined. In the subsequent line, when ```x = 1``` is executed, the variable is defined and assigned a value.

In Python, the type of a variable is determined at the moment it is assigned a value. Therefore, variables do not require explicit declaration or type definition. A variable can hold different types of data without any need for prior declaration.

Consider the example below, where the variable ```x``` takes on different types: ```float```, ```int```, ```bool```, and ```str``` in sequence.

# **5. Other advanced arithmetic operations**


## **a. The `divmod()` function**
- We have previously learned the meaning of the `//` and `%` operators in Python.
- // represents integer division (often denoted as `div` in some other programming languages). `%` represents remainder division (often denoted as mod in other programming languages).
- In reality, in Python: `//` is floor division, which can be defined by the formula: `x // y = floor(x / y)`. `%` is the modulo operator that returns the exact mathematical remainder.
- Consider the following examples to better understand this mechanism:

````python
12.5 // 2
````

````python
12.5 % 2
````

The `divmod()` function in Python can be used to replace both `//` and `%` operations simultaneously. The formula of `divmod()` can be understood as:

`divmod(x, y) = (x // y, x % y).`

````python
divmod(12.5, 2)
````

````python
divmod(12, 3)
````

# **6. Namespace**
We will explore assignment statements in Python in greater depth to understand how they fundamentally differ from those in other programming languages.

The question is: how is an assignment statement, such as `x = 5`, actually executed?

When `x = 5` is executed, the model in other programming languages dictates that the program allocates a memory location for the variable `x` and then places the value `5` into that memory space.

Python handles this in an entirely different manner.\
In Python, all data values—including numbers, strings, and booleans—are true objects (which we will explore later). Each variable has its own name and is managed by Python alongside data objects in an environment called a `Namespace`. This namespace manages the association between `Names (identifiers)` and `Objects`.

---

Each **variable name**, upon executing an assignment statement, is linked to a value object on the other side
```
┌───────────────────┬───────────────────┐
│                   │                   │
│    Names <────────┼──────> Objects    │
│                   │                   │
└───────────────────┴───────────────────┘
```
Each **value is an object**. Value objects can be numbers, strings, booleans, etc.

Within this `Namespace`, whenever an assignment statement is executed, for example: `x = 5`, Python evaluates the value to be assigned in the `Objects` area, then creates a link between the variable x and this numerical value (e.g., 5). This numerical value is called an `lvalue`.

The following diagram illustrates how Python executes the statement `x = 5`. The numeric object evaluates to `lvalue = 5` and creates a link between `x` and this lvalue. Thus, `x` only comes into existence at the time of assignment.

```Python
x = 5
```
```
Namespace
┌───────────────────┬───────────────────┐
│       Names       │      Objects      │
│                   │                   │
│         x ────────┼────────> 5        │
│                   │                   │
└───────────────────┴───────────────────┘
```

Now suppose we proceed to execute the statement `y = x + 2`. The expression on the right-hand side of the assignment is evaluated in the `Objects` area, yielding 7, which becomes `lvalue = 7`. Python then links the variable `y` to this value, meaning `y = 7`. The value of variable `x` remains unchanged.

```Python
x = 5
y = x + 2
```
```
Namespace
┌───────────────────┬───────────────────┐
│       Names       │      Objects      │
│                   │                   │
│         x ────────┼────────> 5        │
│                   │                   │
│         y ────────┼────────> 7        │
│                   │                   │
└───────────────────┴───────────────────┘
```

Next, when executing the statement `x, y = "one"`, `x + 7`, the `Objects` area simultaneously evaluates and creates `"one"` and the value 12 (= x + 7), then concurrently creates the links `x` $\rightarrow$ `"one"` and `y` $\rightarrow$ `12`.

```Python
x = 5
y = x + 2
x, y = "one", x + 7
```
```
Namespace
┌───────────────────┬───────────────────┐
│       Names       │      Objects      │
│                   │                   │
│         x ────┐   │       ┌─> 12      │
│               │   │       │           │
│               └───┼───────┼─> "one"   │
│                   │       │           │
│         y ────────┼───────┘           │
│                   │                   │
└───────────────────┴───────────────────┘
```


````python
# The above sequence of statements is executed in the Python Shell as follows:
x = 5
y = x + 2
x, y = "one", x + 7
x, y
````
