# **3. Help, comment and command seperator**

## **3.1 The ```help``` command**
To view the syntax or meaning of any Python command, we use the ```help()``` function. There are two ways to use it:

**First way:** Type ```help(command_name)``` directly.

````python
help(print)
````

**Second way:** Type ```help()```.
When you do this, the help prompt ```help>``` will appear. You can then type the name of the command you want to get help for after this prompt. For example:

````python
help()
# To exit the help mode, type: quit or Enter on the keyboard
````

The full syntax of the ```print``` command is:
```python
print(value1, value2, ..., valueN, sep=<character or string>, end=<string>, file=<output object>, flush=<True/False>)
```
**Where:**

- ```value1, value2, ..., valueN``` are the values to be printed. These values can be numbers, strings, expressions, or variables.

- ```sep=``` specifies the separator between the values ```value1, value2, ..., valueN```.
The default value is a space character ```' '```.

- ```end=``` specifies the string printed at the end of the output.
The default value is the newline character ```'\n'```, which moves the cursor to the next line.

- The file and flush parameters will be discussed later.


## **3.2 Command seperator**
Python programs are usually written one command per line.

However, in general, it is possible to write multiple commands on the same line. In that case, each command must be separated by a semicolon ```;```.

For example:

````python
print("Hà");print("Nội")
````

In practice, Python programs are typically written with **one command per line**, so **no additional separator is required**. This style improves code readability and is strongly recommended.

##**3.4 Writing Commands Across Multiple Lines**

In Python, a long command can be written across multiple lines.

To continue a command on the next line, use the backslash ```\``` character. After typing ```\```, you can press Enter and continue writing the command on the next line.

````python
a = "Hà Nội là thủ đô \
của Việt Nam"
b = "Hà Nội là thủ đô của việt Nam"
````
