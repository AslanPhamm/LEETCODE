# **2. Explicit type conversion (Type casting)**
To accurately obtain the desired data type from user input, we use Python's built-in type conversion functions: `int()`, `float()`, and `str()`

## **a. The `int()` function**
Syntax: \
`<return_value> = int(<value>)`

This function converts the parameter `<value>` and returns an integer. It only operates successfully when `<value>` is a floating-point number or a string representing an integer. If the argument is a float, `int()` truncates the decimal part and retains only the integer portion.

````python
# For example
int(2.56)
````

````python
int(-15.09)
````

Pay attention to the 3rd and 4th statements in the following example to see that `int()` cannot convert strings that represent floating-point numbers or arithmetic expressions.

````python
int("23")
````

````python
int(23.45)
````

````python
int("12.55")
````

````python
int("34 + 67")
````

## **b. The `float()` function**
Syntax:\
`<return_value> = float(<value>)`

This function converts the parameter `<value>` and returns a floating-point number. It only operates successfully when `<value>` is any numeric expression, or a string representing a float or an integer.

**Note:** The function raises an error if `<value>` is a string containing an arithmetic expression.

In the following examples, observe that the last conversion statement fails because `float()` cannot parse a string containing an arithmetic expression.

````python
float("3.14156")
````

````python
float(5)
````

````python
float(14 + 67 + 3)
````

````python
float("13 + 7.3")
````

To evaluate string expressions, refer to the `eval()` function in the subsequent section.

To convert a string representing a floating-point number into an integer, we must nest the `int()` and `float()` functions together, as shown in the following example:

````python
int(float("3.141567"))
````

## **c. The `str` function**
The `str()` function converts a value or a numeric expression into a string data type `(str)`. Its syntax is similar to the `int()` and `float()` functions introduced earlier:\
``` <return_value> = str(<expression>).```

This function converts any numeric expression into string format. If `<expression>` is already a string, it returns the original value of the string. Note that `<expression>` can be arbitrarily complex.\
See the following examples.

````python
str(5)
````

````python
str(3 + 4**5 - 17.8**2)
````

**Note:** The `str()` function cannot directly handle a series of comma-separated arguments, for example `str(3, 4)`, but it does support converting the tuple result returned by the `divmod()` function.

````python
str(3,4)
````

````python
str(divmod(19,5))
````

# **3. The `eval()` function: Evaluating values from expression strings**

We have learned the purpose of the three functions `int()`, `float()`, and `str()` for explicit type conversion of basic data types, which are widely used when processing input with the `input()` statement. However, another limitation is that these functions cannot handle strings representing mathematical expressions. For example, strings such as `"12 + 5"` or `"23**2 - 12"` cannot be parsed by `int()` or `float()`\
The `eval()` function is precisely what we need in these scenarios.

Syntax: \
`<return_value> = eval(<expression>)`

his function evaluates the expression contained inside a string and returns the corresponding computed value.

````python
eval("12 + 7")
````

````python
eval("24 ** 2 - 15")
````

````python
eval("15**2 + 7.0 + 13 * (34 - 2**1.5)")
````

Consider an introductory example. The following program prompts the user to enter a mathematical expression from the keyboard; once entered, the program evaluates it and immediately outputs the result.

````python
content = """
# Nhập biểu thức từ bàn phím
x = eval(input("Nhập biểu thức toán học bất kỳ: "))
print("Giá trị của biểu thức đã nhập là:", x)
"""

with open("nhapbieuthuc.py", "w") as f:
    f.write(content.strip())
````

````python
!python nhapbieuthuc.py
````

Another unique capability of `eval()` is that it can recognize multiple expressions on a single line separated by commas (`,`).

````python
eval("1+2, 3**2, 14.2 + 16.5**2")
````

````python
m, n = eval("3, 4")
m, n
````

This feature of `eval()` has many practical applications.\

````python
content = """
# Nhập hai hoặc nhiều số trên một hàng
m,n,p = eval(input("Nhập các số m,n,p (cách nhau bởi dấu phẩy): "))
print("Các số đã nhập là:", m, n, p)

"""

with open("nhapbaso.py", "w") as f:
    f.write(content.strip())
````

# **4. The `bool` data type and the `if` statement**
In this section, we will delve deeper into the Boolean (`bool` / logic) data type and the conditional branching `if` statement. With the `if` statement, we will be introduced for the first time to Python's indentation rules.

## **a. The Boolean (`bool`) data type**

The Boolean (`bool`) data type contains only two values: True and False. Boolean data is used to express conditions or comparison operations between numbers or strings.

In Python, the two principal values of this data type are `True` and `False`. The `bool()` function checks whether the value of its argument evaluates to `True` or `False`.

Consider the following examples:

````python
3 > 2
````

````python
x = 2 > 3
bool(x)
````

In the first statement, typing `3 > 2` forms a logical comparison expression, which evaluates to the boolean value `True`.\
In the second statement, the assignment `x = 2 > 3` evaluates the logical expression `2 > 3` to `False`, which is then assigned to the variable `x`, making `x` hold `False`.

Numeric comparison is the simplest way to generate these logical values. The following table describes the syntax for comparison operators in Python.

| Comparison Operator | Meaning | Comparison Question |
| :---: | :--- | :--- |
| `x == y` | Equal to comparison | Is x equal to y? |
| `x != y` | Not equal to comparison | Is x different from y? |
| `x > y` | Greater than comparison | Is x greater than y? |
| `x >= y` | Greater than or equal to comparison | Is x greater than or equal to y? |
| `x < y` | Less than comparison | Is x less than y? |
| `x <= y` | Less than or equal to comparison | Is x less than or equal to y? |

There are three logical operators for the bool data type: `and`, `or`, and `not`. The truth table for these operations is defined as follows:

### Logical `and` Operation

| x | and | y | x and y |
| :---: | :---: | :---: | :---: |
| True | and | True | True |
| False | and | False | False |
| True | and | False | False |
| False | and | True | False |


### Logical `or` Operation

| x | or | y | x or y |
| :---: | :---: | :---: | :---: |
| True | or | True | True |
| False | or | False | False |
| True | or | False | True |
| False | or | True | True |

### Logical `not` Operation

| x | not x |
| :---: | :---: |
| True | False |
| False | True |

## **b. The `if` statement**

The `if` conditional branching statement is one of the most frequently used statements in any programming language. Its practical meaning can be understood through the following real-life scenario:

```
If it stops raining this morning, I will go out; otherwise, I will stay home and study.
```

In the statement above, there are two distinct actions: "go out" and "study". The execution of these actions is not sequential; rather, it depends on the weather condition following the keyword "if". In programming, statements executed conditionally like this are implemented using the `if` statement.

n human language, this statement can be described as:
```
if <condition is true>, then <execute these statements>.

The `if` statement in general, and in Python in particular, has two forms: the short form (single-branch) and the full form (two-branch), as follows:

### Syntax of the short-form `if` statement:
```
if <condition>:
    <statement_block>
```

**Note:**
- A colon : is mandatory after the `<condition>`.
- The entire block of statements belonging to the `if` group must be indented (by at least one space) relative to the `if` keyword.
- All statements within this `if` block must have the exact same level of indentation. This is critical.

**The above statement executes as follows:**\
The program evaluates the `<condition>` following the `if` keyword; if the condition is True, the `<statement_block>` is executed. If the condition is False, the block is bypassed entirely.

### Syntax of the full-form `if` statement:
```
if <condition>:
    <if_statement_block>
else:
    <else_statement_block>
```

**Note:**
- A colon : is required after both the if `<condition>` and the else keyword.
- The entire `<if_statement_block>` and `<else_statement_block>` must be indented by at least one space relative to `if` and `else`.
- All statements within the respective `if` and `else` blocks must share an identical indentation level.

**Execution flow:**\
The program evaluates the `<condition>` following if; if True, the `<if_statement_block>` is executed; if False, the `<else_statement_block>` is executed.

Consider the following examples:\
In the following statement, the `<condition>` evaluates to True, so the `if` block statements are guaranteed to execute.

````python
if True:
  print("Tôi đi chơi đây")
````

The subsequent example is the opposite: because the `<condition>` evaluates to False, the statements will not be executed.

````python
t = 2
if t > 2:
  print(t, "có giá trị lớn hơn 2")
````

The following simple Python program prompts the user to enter two integers, `m` and `n`, and then reports whether they are equal or which one is larger. Observe how nested `if` statements are utilized.

````python
content = """
# Nhập 2 hoặc nhiều số trên một hàng
m,n = eval(input("Nhập các số m,n (cách nhau bởi dấu phẩy): "))
if m == n:
    print("Hai số m, n bằng nhau")
else:
    if m > n:
        print("Số m lớn hơn n")
    else:
        print("Số n lớn hơn m")
"""

with open("sosanh2so.py", "w") as f:
    f.write(content.strip())
````

# **5. The `for` loop**

In Python, we encounter the familiar structure of the `for` statement, similar to the `if` statement discussed earlier. A colon "`:`" is required after the main `for` header line, followed by a newline, and the `<for_statement_block>` must be indented uniformly. Thus, in Python, leading indentation spaces play a vital role.

| Python Syntax | Description |
| :--- | :--- |
| `for k in range(1, 100):` `    <statement_block>` | Iterates `k` over the range from 1 to 99; executes `<statement_block>` on each pass. |
| `for k in range(100, 0, -1):` `    <statement_block>` | Iterates `k` in descending order from 100 down to 1; executes `<statement_block>` on each pass. |

The `range()` function or expression generates an iterable sequence of countable values in Python. This function has several forms, summarized in the table below:

Syntax:
```
range([start], end, [increment])
```

| `range()` Syntax | Description |
| :--- | :--- |
| `range(0, n)` | Sequence of integers from $0, 1, \dots$ up to $n - 1$. |
| `range(1, n, 2)` | Sequence of odd values from $1, 3, \dots$ up to $\le n - 1$, incrementing by 2. |
| `range(n)` | Equivalent to `range(0, n, 1)`: starts at 0, steps by 1 up to $n - 1$. |
| `range(n, 0, -1)` | Decreasing sequence from $n, n - 1, \dots$ down to 1 (step $-1$). |

Let us examine a few examples to better understand the `for` loop statement:

1. Given a natural number n, compute the sum $S = 1 + 2 + \dots + n$. The computation will be performed directly on the Python Shell interactive interface.

````python
n = 100
S = 0
for k in range(1, n + 1):
    S = S + k
print(S)
````

2. The following program prompts the user to enter a natural number n from the keyboard, and then computes all proper divisors of n.

````python
n = int(input("Nhập số tự nhiên n: "))
count = 0
for k in range(1,n):
  if n % k == 0:
    count += 1
print("số các ước số của",n,"là: ",count)
````

# **6. The `while` loop**

Another statement to introduce here is the conditional loop statement `while`. Unlike the `for` statement, which iterates for a predetermined number of iterations, the `while` statement executes an indefinite number of times that depends on its termination condition.

Syntax:
```
while <condition>:
    <loop_statements>
```
A colon `:` is required after the main `while` header statement, followed by a newline, and the `<loop_statements>` must be indented uniformly.

The statement repeatedly executes the `<loop_statements>` in the body of the `while` loop and terminates only based on `<condition>`. Prior to each iteration, the program evaluates `<condition>`; if True, it continues execution; if False, it terminates the loop.

**Example:** Input two natural numbers m and n from the keyboard and compute the Greatest Common Divisor (GCD) of these two numbers.

````python
content = """
M = int(input("Nhập số tự nhiên M: "))
N = int(input("Nhập số tự nhiên N: "))
m = M; n = N
while n != m:
  if n > m:
    n = n - m
  else:
    m = m - n
print("Ước chung lớn nhất của",M,"và",N,"là:",n)
"""

with open("ucln.py", "w") as f:
    f.write(content.strip())
````

````python
!python ucln.py
````

**Note:** In a `while` loop, the `<condition>` is evaluated prior to executing any statements within the loop body (pre-test loop). Therefore, if the `<condition>` evaluates to False right at the beginning, none of the loop statements will be executed.
