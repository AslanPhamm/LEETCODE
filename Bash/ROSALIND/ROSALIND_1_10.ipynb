## Problem 1: A Rapid Introduction to Molecular Biology
**Link:** [https://rosalind.info/problems/dna/](https://rosalind.info/problems/dna/)

A string is simply an ordered collection of symbols selected from some alphabet and formed into a word; the length of a string is the number of symbols that it contains.

An example of a length 21 DNA string (whose alphabet contains the symbols 'A', 'C', 'G', and 'T') is "ATGCTTCAGAAAGGTCTTACG."

**Given**: A DNA string ***s*** of length at most 1000 nt.

**Return**: Four integers (separated by spaces) counting the respective number of times that the symbols 'A', 'C', 'G', and 'T' occur in ***s***

````python
%%bash
cat << 'EOF' > data_bai1.txt
AGCTTTTCATTCTGACTGCAACGGGCAATATGTCTCTGTGTGGATTAAAAAAAGAGTGTCTGATAGCAGC
EOF
````

**Cách 1** Dùng `grep` và `wc -l`

````python
%%bash
# DNA sequence
DNA=$(cat data_bai1.txt)
# Count the occurrences of each nucleotide
A=$(echo "$DNA" | grep -o "A" | wc -l)
T=$(echo "$DNA" | grep -o "T" | wc -l)
C=$(echo "$DNA" | grep -o "C" | wc -l)
G=$(echo "$DNA" | grep -o "G" | wc -l)
# Print the counts
echo "$A $C $G $T"
````

**Cách 2** Dùng `tr` và `paste -sd`\
*Subshell `( )` gom tất cả kết quả lại, `paste -sd` in ket qua trên 1 dòng.*

````python
%%bash
gA=$(cat data_bai1.txt)
(
echo -n $gA | tr -cd 'A' | wc -c
echo -n $gA | tr -cd 'C' | wc -c
echo -n $gA | tr -cd 'G' | wc -c
echo -n $gA | tr -cd 'T' | wc -c
) | paste -sd " "
# dùng paste -sd để gom tất cả lên cùng 1 hàng ngang!
````

**Cách 3** `grep . | sort | uniq -c`

````python
%%bash
DNA=$(cat data_bai1.txt)
# grep -o . sẽ tách mỗi ký tự thành 1 dòng
# sort sẽ xếp A, C, G, T gần nhau
# uniq -c sẽ đếm gộp số lượng của từng nhóm ký tự giống nhau
echo -n "$DNA" | grep -o . | sort | uniq -c
````

## Problem 4: Rabbits and Recurrence Relations
**Link:** [https://rosalind.info/problems/fib/](https://rosalind.info/problems/fib/)\
In the case of Fibonacci's rabbits, any given month will contain the rabbits that were alive the previous month, plus any new offspring. A key observation is that the number of offspring in any month is equal to the number of rabbits that were alive two months prior. As a result, if Fn represents the number of rabbit pairs alive after the (n-th month), then we obtain the Fibonacci sequence having terms Fn that are defined by the recurrence relation `n=F(n−1) + F(n−2)` (with `F1 = F2 = 1` to initiate the sequence).

When finding the n-th term of a sequence defined by a recurrence relation, we can simply use the recurrence relation to generate terms for progressively larger values of ***n***. This problem introduces us to the computational technique of dynamic programming, which successively builds up solutions by using the answers to smaller cases.

**Given:** Positive integers ***n ≤ 40*** and ***k ≤ 5***.

**Return:** The total number of rabbit pairs that will be present after n months, if we begin with 1 pair and in each generation, every pair of reproduction-age rabbits produces a litter of k rabbit pairs (instead of only 1 pair).

----------------------------------------
### SOLUTION 

**Given Parameters:**
* **n** = 5 (Total number of months)
* **k** = 3 (Number of new pairs produced by each adult pair per month)
* **Initial state:** Begin with 1 newborn pair.
* **Growth condition:** Any newborn pair takes 1 month to grow and prepare for reproduction.


**Population Growth Metrics**


| Time (i) | Total Count (F(i)) | Population Definition |
| :--- | :--- | :--- |
| **1** | 1 | 1 (newborn) |
| **2** | 1 | 1 (adult) |
| **3** | 4 | 1 (adult) + 3 (newborn) |
| **4** | 7 | 4 (adult) + 3 (newborn) |
| **5** | 19 | 7 (adult) + 12 (newborn) |


**Summary Formula**

The sequence follows a generalized Fibonacci recurrence relation. To calculate the total number of pairs at any given month *i*, use the following formula:

**F(i) = F(i-1) + k * F(i-2)**

**Where:**
* **F(i)** is the total population at month *i*.
* **F(i-1)** represents the total population from the previous month (which all survive to the current month).
* **k * F(i-2)** represents the new offspring (the population from two months ago is now fully mature and each pair produces *k* new pairs).

**Cách 1**

````python
%%bash
n=5
k=3
# Assign the varriables
Fibo[1]=1
Fibo[2]=1
# Use for loop to print the first n Fibonacci numbers
for ((i=3; i<=n; i++)); do
# Conducting the calculation
    Fibo[$i]=$((Fibo[i-1] + k*Fibo[i-2]))    
    echo "${Fibo[$n]}"
done
````

**Cách 2** Su dung `awk`

````python
%%bash
echo "5 3" | \
awk '{n=$1; k=$2; f1=1; f2=1; \
    for(i=3;i<=n;i++){f3=f2+f1*k; f1=f2; f2=f3} \
    print f2}'
````

## Problem 5: Computing GC Content 
**Link:** [https://rosalind.info/problems/gc/](https://rosalind.info/problems/gc/)\
The GC-content of a DNA string is given by the percentage of symbols in the string that are 'C' or 'G'. For example, the GC-content of "AGCTATAG" is 37.5%. Note that the reverse complement of any DNA string has the same GC-content.

DNA strings must be labeled when they are consolidated into a database. A commonly used method of string labeling is called FASTA format. In this format, the string is introduced by a line that begins with '>', followed by some labeling information. Subsequent lines contain the string itself; the first line to begin with '>' indicates the label of the next string.

In Rosalind's implementation, a string in FASTA format will be labeled by the ID "Rosalind_xxxx", where "xxxx" denotes a four-digit code between 0000 and 9999.

**Given**: At most 10 DNA strings in FASTA format (of length at most 1 kbp each).

**Return**: The ID of the string having the highest GC-content, followed by the GC-content of that string. Rosalind allows for a default error of 0.001 in all decimal answers unless otherwise stated; please see the note on absolute error below.

````python
%%bash
# Dữ liệu dùng chung cho Bài 5 (Định dạng FASTA)
cat << 'EOF' > data_bai5.txt
>Rosalind_6404
CCTGCGGAAGATCGGCACTAGAATAGCCAGAACCGTTTCTCTGAGGCTTCCGGCCTTCCC
TCCCACTAATAATTCTGAGG
>Rosalind_5959
CCATCGGTAGCGCATCCTTAGTCCAATTAAGTCCCTATCCAGGCGCTCCGCCGAAGGTCT
ATATCCATTTGTCAGCAGACACGC
>Rosalind_0808
CCACCCTCGTGGTATGGCTAGGCATTCAGGAACCGGAGAACGCTTCAGACCAGCCCGGAC
TGGGAACCTGCGGGCAGTAGGTGGAAT
EOF
````

**Cách 1** Tính thủ công từng chuỗi

````python
%%bash
FASTA_DATA=$(cat data_bai5.txt)

# 1. Assign variables
MAX_GC=0
MAX_HEADER=""

# 2. While loop
while IFS=$'\t' read -r header sequence; do

# 3. Adjust the sequence to be in 1 line
    seq=$(echo -n "$sequence" | tr -d '\n')

# 4. Calculate the GC content
    gc_count=$(echo -n "$seq" | tr -cd 'GC' | wc -m)
    total_count=${#seq}
    gc_pct=$(echo "scale=6; ($gc_count/$total_count) * 100" | bc)

# 5. Compare the GC content to find the maximum
    if (( $(echo "$gc_pct > $MAX_GC" | bc -l) )); then
        MAX_GC=$gc_pct
        MAX_HEADER=$header
    fi

done < <(echo "$FASTA_DATA" | awk '/>/ {printf("%s%s\t",(N==0?"":"\n"),$0); N++}
                                    !/>/ {printf("%s",$0)}
                                    END {print ""}')

echo "${MAX_HEADER#>}"
echo "$MAX_GC"
````

**Cách 2** Dùng `awk`\
*Hàm `gsub` (Global Substitution) đếm lượng G/C.*

````python
%%bash
awk '
BEGIN {n=0; total=0;}
NR==1 {print; next}
/>/   { 
        print n*100/total "\n" $0;
        n=0; total=0; next
      }
      {
        n += gsub("[GC]", "&");
        total += gsub("[ATCG]", "&")
      }
END   {print n*100/total}
' data_bai5.txt
````

## Problem 6: Counting Point Mutations
**Link:** [https://rosalind.info/problems/hamm/](https://rosalind.info/problems/hamm/)\
Given two strings ***s*** and ***t*** of equal length, the Hamming distance between ***s*** and ***t***, denoted dH(***s,t***), is the number of corresponding symbols that differ in ***s*** and ***t***.\
![image.png](./DATA/HAMM.png)

**Given:** Two DNA strings **s** and **t** of equal length (not exceeding 1 kbp).

**Return:** The Hamming distance dH(***s,t***).

---
### SOLUTION
- **Compare Characters:** At each position, extract and compare the character from string s with the character at the exact same position in string t.
- **Increment Mismatches:** If the two characters are different (a mismatch), increase the mutation counter by 1. If they are identical, do nothing and proceed.

````python
%%bash
# Dữ liệu dùng chung cho Bài 6
cat << 'EOF' > data_bai6.txt
GAGCCTACTAACGGGAT
CATCGTAATGACGGCCT
EOF
````

**Cách 1** Bash Script\
*dùng vòng lặp For và cắt chuỗi (`${s:$i:1}`)*

````python
%%bash
t=$(head -n 1 data_bai6.txt)
s=$(tail -n 1 data_bai6.txt)
# Calculate length of the DNA sequence
len=${#s}
# Find the mutation in the DNA sequence when comarring with the reference
hamming_dist=0
# Use for loop
for (( i=0; i<len; i+=1 )) # from i=0 increase 1 to the length of the DNA
do
    # Get the character (1) at position i in both strings
    si=${s:$i:1}
    ti=${t:$i:1}
    # If si != ti, increase haming_dist by 1
    if [ "$si" != "$ti" ]; then
        hamming_dist=$((hamming_dist + 1))
    fi
done
# Print
echo "$hamming_dist"
````

**Cách 2** Dùng `awk` và `substr`\
*OPTION `-v` truyền 2 biến từ bên ngoài vào trong `awk`.*

````python
!awk \
-v t=$(head -n 1 data_bai6.txt) \
-v s=$(tail -n 1 data_bai6.txt) \
'BEGIN { hd = 0; for (i=1; i<=length(s); i++) \
    { if (substr(s, i, 1) != substr(t, i, 1)) { hd++ } } \
    print hd }'
````

**Cách 3** Dùng `awk` đọc mảng\
*Dùng hàm `split` để chia nhỏ chuỗi thành một mảng các ký tự riêng biệt rồi so sánh từng Index.*

````python
!awk 'NR==1 { split($0, s, "") } NR==2 { split($0, t, "");\
     for(i=1; i<=length($0); i++) if(s[i] != t[i]) count++; print count }'\
     data_bai6.txt
````

**Cách 4** Dung lệnh **`cmp`** (Compare)

````python
%%bash
s=$(head -n 1 data_bai6.txt)
t=$(tail -n 1 data_bai6.txt)

# 1. Cú pháp <(...) gọi là Process Substitution. Nó giả lập chuỗi bộ nhớ thành 1 file ảo để truyền vào cmp.
# 2. Tùy chọn -l (list) sẽ liệt kê ra toàn bộ các byte bị khác biệt.
# 3. Dùng lệnh wc -l để đếm xem có bao nhiêu dòng khác biệt được liệt kê ra.
cmp -l <(echo -n "$s") <(echo -n "$t") | wc -l
````

## Problem 7: Mendel's First Law
**Link:** [https://rosalind.info/problems/iprb/](https://rosalind.info/problems/iprb/)\
**Given:** Three positive integers ***k***, ***m***, and ***n***, representing a population containing ***k+m+n*** organisms: \
    - ***k*** individuals are homozygous dominant for a factor, \
    - ***m*** are heterozygous, \
    - and ***n*** are homozygous recessive.

**Return:** The probability that two randomly selected mating organisms will produce an individual possessing a dominant allele (and thus displaying the dominant phenotype). Assume that any two organisms can mate.



**Cách 1** Tính xác xuất của từng nhóm và cổng chúng lại

````python
%%bash
#k= AA, m= Aa, n= aa
k=2
m=2
n=2

awk -v k="$k" -v m="$m" -v n="$n" 'BEGIN {
    total = k + m + n;

    # 1. Xác suất chọn được 2 cá thể đồng hợp lặn (aa x aa) -> sinh 100% con aa
    p_nn = (n / total) * ((n - 1) / (total - 1));

    # 2. Xác suất chọn được 2 cá thể dị hợp (Aa x Aa) -> sinh 25% con aa
    p_mm = (m / total) * ((m - 1) / (total - 1)) * 0.25;

    # 3. Xác suất chọn được 1 dị hợp và 1 đồng hợp lặn (Aa x aa hoặc aa x Aa) -> sinh 50% con aa
    p_mn = ((m / total) * (n / (total - 1)) + (n / total) * (m / (total - 1))) * 0.5;

    # Tổng xác suất sinh ra con mang kiểu hình lặn (aa)
    p_recessive = p_nn + p_mm + p_mn;

    # Xác suất sinh ra con mang kiểu hình trội (phần bù)
    p_dominant = 1 - p_recessive;

    # Tỉ lệ có đồng hợp tử
    printf p_dominant;
}'
````

**Cách 2** Dung lệnh `bc`\
*Lệnh `bc` (Basic Calculator) của Linux dung để tính*

````python
%%bash
k=2
m=2
n=2
N=$((k+m+n))
echo "scale=5; 1 - ($n*($n-1) + ($m*($m-1))*0.25 + $m*$n*2*0.5) / ($N*($N-1))" | bc
````

**Cách 3** \
*Biểu thức `$m*$n*2*0.5` trong cách 2 thực chất bị triệt tiêu đi số `2` và `0.5`, chỉ còn lại `$m*$n`.*

````python
%%bash
echo "2 2 2" | awk '{k=$1; m=$2; n=$3; N=k+m+n; total=N*(N-1);
     aa=n*(n-1)+(m*n)+0.25*m*(m-1);
     printf "%.5f\n", 1-(aa/total)}'
````

The 20 commonly occurring amino acids are abbreviated by using 20 letters from the English *alphabet* (all letters except for B, J, O, U, X, and Z). *Protein strings* are constructed from these 20 symbols. Henceforth, the term *genetic string* will incorporate protein strings along with *DNA strings* and *RNA strings*.

The RNA codon table dictates the details regarding the encoding of specific codons into the amino acid alphabet.

**Given:** An RNA string s
 corresponding to a strand of mRNA (of length at most 10 *kbp*).

**Return:** The protein string encoded by **s**.

````python
%%bash
# Dữ liệu mẫu dùng chung cho Bài 8
cat << 'EOF' > data_bai8.txt
AUGGCCAUGGCGCCCAGAACUGAGAUCAAUAGUACCCGUAUUAACGGGUGA
EOF
````

**Cách 1** Hàm `translate` đã cài sẵn\
*`fold -w 3` chia chuỗi ra từng cụm 3 ký tự (Codon), sau đó `sed -e` translate codon to aa*

````python
%%bash
# Các hàm dùng chung (VD: hàm dịch mã protein)
mkdir -p "$HOME/.local/bin"

cat <<'EOF' > "$HOME/.local/bin/translate"
#!/bin/bash
input="${1:-$(cat)}"
printf '%s' "$input" | fold -w 3 | awk 'length($0) == 3' | sed \
    -e 's/UUU/F/' -e 's/UUC/F/' -e 's/UUA/L/' -e 's/UUG/L/' \
    -e 's/CUU/L/' -e 's/CUC/L/' -e 's/CUA/L/' -e 's/CUG/L/' \
    -e 's/AUU/I/' -e 's/AUC/I/' -e 's/AUA/I/' -e 's/AUG/M/' \
    -e 's/GUU/V/' -e 's/GUC/V/' -e 's/GUA/V/' -e 's/GUG/V/' \
    -e 's/UCU/S/' -e 's/UCC/S/' -e 's/UCA/S/' -e 's/UCG/S/' \
    -e 's/CCU/P/' -e 's/CCC/P/' -e 's/CCA/P/' -e 's/CCG/P/' \
    -e 's/ACU/T/' -e 's/ACC/T/' -e 's/ACA/T/' -e 's/ACG/T/' \
    -e 's/GCU/A/' -e 's/GCC/A/' -e 's/GCA/A/' -e 's/GCG/A/' \
    -e 's/UAU/Y/' -e 's/UAC/Y/' -e 's/UAA/*/' -e 's/UAG/*/' \
    -e 's/CAU/H/' -e 's/CAC/H/' -e 's/CAA/Q/' -e 's/CAG/Q/' \
    -e 's/AAU/N/' -e 's/AAC/N/' -e 's/AAA/K/' -e 's/AAG/K/' \
    -e 's/GAU/D/' -e 's/GAC/D/' -e 's/GAA/E/' -e 's/GAG/E/' \
    -e 's/UGU/C/' -e 's/UGC/C/' -e 's/UGA/*/' -e 's/UGG/W/' \
    -e 's/CGU/R/' -e 's/CGC/R/' -e 's/CGA/R/' -e 's/CGG/R/' \
    -e 's/AGU/S/' -e 's/AGC/S/' -e 's/AGA/R/' -e 's/AGG/R/' \
    -e 's/GGU/G/' -e 's/GGC/G/' -e 's/GGA/G/' -e 's/GGG/G/' \
    | tr -d '\n' | cut -d '*' -f1
printf '\n'
EOF

chmod +x "$HOME/.local/bin/translate"
export PATH="$HOME/.local/bin:$PATH"
````

````python
%%bash
rna=$(cat data_bai8.txt)
translate "$rna"
````

**Cách 2** Dùng Associative Array\
***Mảng kết hợp** (Associative Array - giống như Hash Map/Từ điển) by lệnh `declare -A`. Ta có thể ánh xạ Codon -> Protein*

````python
%%bash
# Khai báo mảng kết hợp (Dictionary)
declare -A codon2aa=(
    [UUU]=F [UUC]=F [UUA]=L [UUG]=L [UCU]=S [UCC]=S [UCA]=S [UCG]=S
    [UAU]=Y [UAC]=Y [UAA]=* [UAG]=* [UGU]=C [UGC]=C [UGA]=* [UGG]=W
    [CUU]=L [CUC]=L [CUA]=L [CUG]=L [CCU]=P [CCC]=P [CCA]=P [CCG]=P
    [CAU]=H [CAC]=H [CAA]=Q [CAG]=Q [CGU]=R [CGC]=R [CGA]=R [CGG]=R
    [AUU]=I [AUC]=I [AUA]=I [AUG]=M [ACU]=T [ACC]=T [ACA]=T [ACG]=T
    [AAU]=N [AAC]=N [AAA]=K [AAG]=K [AGU]=S [AGC]=S [AGA]=R [AGG]=R
    [GUU]=V [GUC]=V [GUA]=V [GUG]=V [GCU]=A [GCC]=A [GCA]=A [GCG]=A
    [GAU]=D [GAC]=D [GAA]=E [GAG]=E [GGU]=G [GGC]=G [GGA]=G [GGG]=G
)

rna=$(cat data_bai8.txt)
protein=""

for (( i=0; i<${#rna}; i+=3 )); do
    codon=${rna:$i:3}          # Cắt 3 ký tự hiện tại
    aa=${codon2aa[$codon]}     # Tra codon sang amino acid

    protein+="$aa"
done

echo "$protein"
````

## Problem 9: Finding a Motif in DNA
**Link:** [https://rosalind.info/problems/subs/](https://rosalind.info/problems/subs/)\
Given two strings $s$ and $t$, $t$ is a **substring** of $s$ if $t$ is contained as a contiguous collection of symbols in $s$ (as a result, $t$ must be no longer than $s$).\
The **position** of a symbol in a string is the total number of symbols found to its left, including itself (e.g., the positions of all occurrences of `'U'` in `"AUGCUUCAGAAAGGUCUUACG"` are $2, 5, 6, 15, 17,$ and $18$). The symbol at position $i$ of $s$ is denoted by $s[i]$.\
A substring of $s$ can be represented as $s[j:k]$, where $j$ and $k$ represent the starting and ending positions of the substring in $s$; for example, if $s = \text{"AUGCUUCAGAAAGGUCUUACG"}$, then $s[2:5] = \text{"UGCU"}$.\
The **location** of a substring $s[j:k]$ is its beginning position $j$; note that $t$ will have multiple locations in $s$ if it occurs more than once as a substring of $s$ (see the Sample below).

**Given:** Two DNA strings $s$ and $t$ (each of length at most 1 kbp).

**Return:** All locations of $t$ as a substring of $s$.

## Problem 10: Consensus and Profile
**Link:** [https://rosalind.info/problems/cons/](https://rosalind.info/problems/cons/)\
A matrix is a rectangular table of values divided into rows and columns. An $m \times n$ matrix has $m$ rows and $n$ columns. Given a matrix $A$, we write $A_{i,j}$ to indicate the value found at the intersection of row $i$ and column $j$.\
Say that we have a collection of DNA strings, all having the same length $n$. Their **profile matrix** is a $4 \times n$ matrix $P$ in which $P_{1,j}$ represents the number of times that `'A'` occurs in the $j$-th position of one of the strings, $P_{2,j}$ represents the number of times that `'C'` occurs in the $j$-th position, and so on (see below).\
A **consensus string** $c$ is a string of length $n$ formed from our collection by taking the most common symbol at each position; the $j$-th symbol of $c$ therefore corresponds to the symbol having the maximum value in the $j$-th column of the profile matrix. Of course, there may be more than one most common symbol, leading to multiple possible consensus strings.

**Example Matrix Representation**
```text
DNA Strings:
A T C C A G C T
G G G C A A C T
A T G G A T C T
A A G C A A C C
T T G G A A C T
A T G C C A T T
A T G G C A C T

Profile:
A:  5 1 0 0 5 5 0 0
C:  0 0 1 4 2 0 6 1
G:  1 1 6 3 0 1 0 0
T:  1 5 0 0 0 1 1 6

Consensus:
A T G C A A C T
```

**Given:** A collection of at most 10 DNA strings of equal length (at most 1 kbp) in FASTA format.\
**Return:** A consensus string and profile matrix for the collection. (If several possible consensus strings exist, then you may return any one of them.)

````python
%%bash
# Dữ liệu dùng chung cho Bài 10 (FASTA Format)
cat << 'EOF' > data_bai10.txt
>Rosalind_1
ATCCAGCT
>Rosalind_2
GGGCAACT
>Rosalind_3
ATGGATCT
>Rosalind_4
AAGCAACC
>Rosalind_5
TTGGAACT
>Rosalind_6
ATGCCATT
>Rosalind_7
ATGGCACT
EOF
````

**Cách 1** Duyệt CỘT trước, CHUỖI sau

````python
%%bash
FILE=data_bai10.txt

# Đọc sequences vào mảng (bỏ header)
seqs=( $(awk '!/^>/' "$FILE") )

# Số chuỗi và chiều dài
num_seqs=${#seqs[@]}
seq_len=${#seqs[0]}

seqA=()
seqC=()
seqG=()
seqT=()

# Vòng NGOÀI: từng CỘT
for (( j=0; j<seq_len; j++ )); do
    A=0; C=0; G=0; T=0

    # Vòng TRONG: từng CHUỖI
    for (( i=0; i<num_seqs; i++ )); do
        nu=${seqs[$i]:$j:1}

        if [[ "$nu" == "A" ]]; then (( A++ )); fi
        if [[ "$nu" == "C" ]]; then (( C++ )); fi
        if [[ "$nu" == "G" ]]; then (( G++ )); fi
        if [[ "$nu" == "T" ]]; then (( T++ )); fi
    done

    seqA+=("$A")
    seqC+=("$C")
    seqG+=("$G")
    seqT+=("$T")
done

# Tìm consensus: mỗi cột chọn nucleotide có count lớn nhất
consensus=""
for (( j=0; j<seq_len; j++ )); do
    max=${seqA[$j]}; best="A"
    if (( seqC[$j] > max )); then max=${seqC[$j]}; best="C"; fi
    if (( seqG[$j] > max )); then max=${seqG[$j]}; best="G"; fi
    if (( seqT[$j] > max )); then max=${seqT[$j]}; best="T"; fi
    consensus+="$best"
done

# In kết quả
echo "$consensus"
echo "A: ${seqA[*]}"
echo "C: ${seqC[*]}"
echo "G: ${seqG[*]}"
echo "T: ${seqT[*]}"
````
